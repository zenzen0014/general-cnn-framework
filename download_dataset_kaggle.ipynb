```python
import shutil
from pathlib import Path
import kagglehub

# Download ke cache KaggleHub
path = kagglehub.dataset_download(
    "barkataliarbab/banana-ripeness-classification-classification"
)

# Copy hasil download ke folder project
target_dir = Path("./dataset/banana-ripeness").resolve()

if target_dir.exists():
    shutil.rmtree(target_dir)

shutil.copytree(path, target_dir)

print("Dataset copied to:", target_dir)
```
